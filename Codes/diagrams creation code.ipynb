```python
from PIL import Image
import matplotlib.pyplot as plt
import os

# ---------------------------
# User settings (edit if needed)
# ---------------------------
folder = r"E:\Abroad period research\Time-Series KD + Explainability\Multimodel codes\TReX_Publication_Results\RonsadelVAalle\Visualizations"

filenames = [
    "SHAP_Summary_Regime_0_RonsadelVAalle.png",
    "SHAP_Summary_Regime_1_RonsadelVAalle.png",
    "SHAP_Summary_Regime_2_RonsadelVAalle.png",
    "SHAP_Summary_Regime_3_RonsadelVAalle.png",
    "SHAP_Summary_Regime_4_RonsadelVAalle.png"
]

# Output filenames
output_png = os.path.join(folder, "SHAP_Summary_3x2_Combined.png")
output_tiff = os.path.join(folder, "SHAP_Summary_3x2_Combined.tiff")

# Quality settings
DPI = 600                      # output resolution
panels_per_row = 2
rows = 3
# Target pixel width for each panel to make text readable (adjust if you want larger)
target_panel_width_px = 2500   # e.g., 2500 px wide per panel
# ---------------------------

# Build full paths and check existence
image_paths = []
missing = False
for f in filenames:
    full_path = os.path.join(folder, f)
    if not os.path.exists(full_path):
        print(f"ERROR: File not found: {full_path}")
        missing = True
    else:
        image_paths.append(full_path)

if missing:
    raise SystemExit("One or more image files are missing. Fix paths/names and rerun.")

# Load images and upsample if needed
images = []
for p in image_paths:
    img = Image.open(p).convert("RGBA")  # keep transparency if present
    # If image width is smaller than target, upscale to improve text clarity
    if img.width < target_panel_width_px:
        scale = target_panel_width_px / img.width
        new_w = int(img.width * scale)
        new_h = int(img.height * scale)
        img = img.resize((new_w, new_h), Image.LANCZOS)
        print(f"Upscaled {os.path.basename(p)} to {new_w}x{new_h}px")
    else:
        print(f"No upscale needed for {os.path.basename(p)} ({img.width}x{img.height}px)")
    images.append(img)

# If fewer images than grid cells, append blank images for empty slots
total_cells = rows * panels_per_row
while len(images) < total_cells:
    # Create a white blank image matching median size of your images
    median_w = int(sum(im.width for im in images) / len(images))
    median_h = int(sum(im.height for im in images) / len(images))
    blank = Image.new("RGBA", (median_w, median_h), (255,255,255,255))
    images.append(blank)

# Create figure
# Figure size in inches = (panel_width_inches * panels_per_row, panel_height_inches * rows)
# We compute panel_width_inches from target_panel_width_px / DPI
panel_width_in = target_panel_width_px / DPI
# Estimate panel height from the average aspect ratio
avg_aspect = sum(im.height/im.width for im in images[:len(filenames)]) / len(filenames)
panel_height_in = panel_width_in * avg_aspect

fig_width = panel_width_in * panels_per_row
fig_height = panel_height_in * rows

fig, axes = plt.subplots(rows, panels_per_row, figsize=(fig_width, fig_height), dpi=DPI)
axes = axes.flatten()

# Plot images
for ax, img in zip(axes, images):
    # Convert to array-friendly RGB for matplotlib
    if img.mode != "RGB":
        img_plot = img.convert("RGB")
    else:
        img_plot = img
    ax.imshow(img_plot)
    ax.axis("off")

plt.tight_layout(pad=0.5)

# Save high-resolution outputs
print("Saving PNG (this may take a moment)...")
plt.savefig(output_png, dpi=DPI, bbox_inches='tight', pad_inches=0.1)
print(f"Saved: {output_png}")

print("Saving TIFF (this may take longer)...")
plt.savefig(output_tiff, dpi=DPI, bbox_inches='tight', pad_inches=0.1, format='tiff')
print(f"Saved: {output_tiff}")

plt.close(fig)
print("Done.")
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import os
import numpy as np

# ---------------------------
# User settings (edit if needed)
# ---------------------------
folder = r"E:\Abroad period research\Time-Series KD + Explainability\Multimodel codes\TReX_Publication_Results\RonsadelVAalle\Visualizations"

filenames = [
    "SHAP_Summary_Regime_0_RonsadelVAalle.png",
    "SHAP_Summary_Regime_1_RonsadelVAalle.png",
    "SHAP_Summary_Regime_2_RonsadelVAalle.png",
    "SHAP_Summary_Regime_3_RonsadelVAalle.png",
    "SHAP_Summary_Regime_4_RonsadelVAalle.png"
]

output_png = os.path.join(folder, "SHAP_Summary_3x2_Centered_Combined.png")
output_tiff = os.path.join(folder, "SHAP_Summary_3x2_Centered_Combined.tiff")

# Quality settings
DPI = 600                      # output resolution
rows = 3
cols = 2
# Increase target panel width to make small text readable
target_panel_width_px = 3000   # try 3000 px; raise if text still small
upsample_filter = Image.LANCZOS
# ---------------------------

# Build full paths and check existence
image_paths = []
missing = False
for f in filenames:
    full_path = os.path.join(folder, f)
    if not os.path.exists(full_path):
        print(f"ERROR: File not found: {full_path}")
        missing = True
    else:
        image_paths.append(full_path)

if missing:
    raise SystemExit("One or more image files are missing. Fix paths/names and rerun.")

# Load images and upsample if needed
images = []
for p in image_paths:
    img = Image.open(p).convert("RGB")
    if img.width < target_panel_width_px:
        scale = target_panel_width_px / img.width
        new_w = int(img.width * scale)
        new_h = int(img.height * scale)
        img = img.resize((new_w, new_h), upsample_filter)
        print(f"Upscaled {os.path.basename(p)} to {new_w}x{new_h}px")
    else:
        print(f"No upscale needed for {os.path.basename(p)} ({img.width}x{img.height}px)")
    images.append(img)

# Create blank image if needed (not expected here)
total_cells = rows * cols
if len(images) < total_cells - 1:  # we need at least 5 images for layout
    median_w = int(np.median([im.width for im in images]))
    median_h = int(np.median([im.height for im in images]))
    while len(images) < 5:
        images.append(Image.new("RGB", (median_w, median_h), (255,255,255)))

# Compute figure size based on target panel width and average aspect ratio
panel_width_in = target_panel_width_px / DPI
# For height, use the median aspect ratio of the first four images (grid panels)
aspect_ratios = [im.height / im.width for im in images[:4]]
avg_aspect = float(np.median(aspect_ratios))
panel_height_in = panel_width_in * avg_aspect

fig_width = panel_width_in * cols
# For the bottom row, we want the image to be slightly taller; use same panel_height_in
fig_height = panel_height_in * rows

# Create figure with GridSpec so last image spans two columns
fig = plt.figure(figsize=(fig_width, fig_height), dpi=DPI)
gs = gridspec.GridSpec(rows, cols, figure=fig, hspace=0.3, wspace=0.25)

# Place images: first four fill the 2x2 grid, fifth spans both columns on row 3
ax_positions = [
    gs[0, 0],
    gs[0, 1],
    gs[1, 0],
    gs[1, 1],
    gs[2, 0:2]  # span both columns
]

for ax_pos, img in zip(ax_positions, images[:5]):
    ax = fig.add_subplot(ax_pos)
    # center image within the axes and preserve aspect
    ax.imshow(img)
    ax.set_xticks([])
    ax.set_yticks([])
    # remove axis frame
    for spine in ax.spines.values():
        spine.set_visible(False)

# If there are extra axes (not expected), hide them
all_axes = fig.get_axes()
for ax in all_axes:
    ax.set_axis_off()

# Tight layout and save
plt.tight_layout()
print("Saving PNG (high-res)...")
plt.savefig(output_png, dpi=DPI, bbox_inches='tight', pad_inches=0.1)
print(f"Saved: {output_png}")

print("Saving TIFF (high-res)...")
plt.savefig(output_tiff, dpi=DPI, bbox_inches='tight', pad_inches=0.1, format='tiff')
print(f"Saved: {output_tiff}")

plt.close(fig)
print("Done.")
```
